```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import os
```

```python
ordner = "/Users/simonhein/Documents/Uni Würzburg/WiSe 2025:26/Projektarbeit/Python/Erstes Skript/Testdaten"
dateien = os.listdir(ordner)


for datei in dateien:

    #Nr. der Datei rausfinden:
    res = ""
    for ch in datei:
        if ch.isdigit():
            res = res + ch


    pfad = f"{ordner}/{datei}"
    data = pd.read_excel(pfad, skiprows=4)

    #Parameter definieren:
    wavelength = data.iloc[:,0]

    intensity = data.iloc[:,1:]

    water = data.iloc[:,1]

    time = np.arange(len(data.columns)-1)


    #Spektrum bereinigen (Wasser entfernen)
    for col in intensity.columns:
        intensity[col] = intensity[col] - water


    #Flip:
    intensity = intensity.mul(-1)


    #Maximale Intensität finden und intensity normieren
    maxint = max(intensity.max(axis=1))
    intensity = intensity / maxint


    #Plot:
    time_grid, wavelength_grid = np.meshgrid(time, wavelength)

    fig = go.Figure()

    fig.add_trace(go.Surface(
        x=wavelength_grid,  # X-axis: Wavelength
        y=time_grid,        # Y-axis: Time
        z=intensity,      # Z-axis: Intensity
        colorscale="Turbo",  # Color scale for intensity
        #colorbar=dict(title="Intensity")  # Optional: Add color bar
    ))

    fig.update_layout(
    # title="3D Plot",
        scene=dict(
            xaxis_title="Wavelength [λ]",
            yaxis_title="Time [s]",
            zaxis_title="Absorbance [a.u.]",

            
            aspectratio=dict(
                x=1,  # Keep X-axis as is
                y=1,  # Stretch the Y-axis by increasing its aspect ratio
                z=1   # Keep Z-axis as is
            )
        )
    )

    #fig.show()


    #Plot speichern:
    fig.write_html(f"/Users/simonhein/Documents/Uni Würzburg/WiSe 2025:26/Projektarbeit/Python/Erstes Skript/Plots/{res}_plot.html")


    #bereinigte Tabelle erstellen und speichern:
    newdata = pd.concat([wavelength, intensity], axis=1)
    newdata.to_excel(f"/Users/simonhein/Documents/Uni Würzburg/WiSe 2025:26/Projektarbeit/Python/Erstes Skript/Testdaten_bereinigt/{res}_bereinigt.xlsx")



```
